### 1.5 Key Performance Indicators (KPIs)

The following KPIs will be used to evaluate Superstore's business performance:

| KPI | Formula | Purpose |
|---|---|---|
| Total Sales | Sum of Sales | Measures total revenue generated |
| Total Profit | Sum of Profit | Measures overall profitability |
| Total Orders | Count of unique Order IDs | Measures the number of distinct customer orders |
| Total Quantity Sold | Sum of Quantity | Measures the total number of units sold |
| Average Order Value (AOV) | Total Sales / Total Orders | Measures the average revenue generated per order |
| Overall Profit Margin | Total Profit / Total Sales × 100 | Measures how much profit is generated from each unit of sales |
| Average Discount | Sum of Discount / Number of transaction lines | Measures the average discount applied across transaction lines |
| Sales Growth Rate | (Current Period Sales - Previous Period Sales) / Previous Period Sales × 100 | Measures the percentage change in sales between periods |
| Profit Growth Rate | (Current Period Profit - Previous Period Profit) / Previous Period Profit × 100 | Measures the percentage change in profit between periods |
| Average Shipping Time | Sum of Shipping Days / Number of Orders | Measures the average time between order and shipment dates |
| Loss-Making Transaction Rate | Number of loss-making transaction lines / Total transaction lines × 100 | Measures the proportion of transaction lines generating losses |

### 3.1 Create a Working Copy

A separate copy of the original DataFrame is created for data cleaning and transformation. This preserves the original imported dataset and allows all cleaning operations to be performed without modifying the raw data held in memory.

### 3.2 Standardize Column Names

Column names were standardized to improve consistency and make them easier to reference in Python and SQL.

The following transformations were applied:

- Leading and trailing spaces were removed.
- Column names were converted to lowercase.
- Spaces were replaced with underscores.
- Hyphens were replaced with underscores.

For example, `Order ID` was converted to `order_id`, while `Sub-Category` was converted to `sub_category`.

```python
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)
```

```python
df_clean.columns
```

### 3.3 Convert Date Columns

The `order_date` and `ship_date` columns were converted from object (text) data types to datetime format.

This conversion allows the dataset to support time-based analysis and date calculations, including:

- Sales and profit trends over time.
- Year and month analysis.
- Shipping processing time calculations.
- Year-over-year performance comparisons.

The converted date columns were verified to confirm that they are now stored as datetime values.

```python
df_clean["order_date"] = pd.to_datetime(df_clean["order_date"])
df_clean["ship_date"] = pd.to_datetime(df_clean["ship_date"])
```

```python
df_clean[["order_date", "ship_date"]].dtypes
```

### 3.4 Create Date-Based Columns

New date-based columns were created from `order_date` to support time-based analysis.

- `order_year` identifies the year in which an order was placed.
- `order_month` contains the name of the order month.
- `month_year` represents the specific year and month of each order.

These derived columns will support yearly performance analysis, monthly trend analysis, seasonality analysis, and year-over-year comparisons.

```python
df_clean["order_year"] = df_clean["order_date"].dt.year
df_clean["order_month"] = df_clean["order_date"].dt.month_name()
df_clean["month_year"] = df_clean["order_date"].dt.to_period("M")
```

```python
df_clean[
    ["order_date", "order_year", "order_month", "month_year"]
].head()
```

### 3.5 Create Shipping Processing Days

A new column called `shipping_days` was created by calculating the difference between `ship_date` and `order_date`.

This column represents the number of days between when an order was placed and when it was shipped.

The new variable will support analysis of:

- Average order-to-shipment processing time.
- Differences in processing time across shipping modes.
- Orders with unusually long or short processing times.

The calculated values were checked to confirm that no negative shipping times were present.

**Note:** `shipping_days` represents order-to-shipment processing time and not total delivery time, because the dataset does not contain the actual customer delivery date.

```python
df_clean["shipping_days"] = (
    df_clean["ship_date"] - df_clean["order_date"]
).dt.days
```

```python
df_clean[
    ["order_date", "ship_date", "ship_mode", "shipping_days"]
].head(10)
```

### 3.6 Create Profit Margin

A new column called `profit_margin` was created by dividing Profit by Sales for each transaction line.

Profit margin measures the profit or loss generated relative to the amount of sales.

- A positive profit margin represents a profitable transaction.
- A negative profit margin represents a loss-making transaction.
- A higher positive margin indicates greater profitability relative to sales.

The transaction-level profit margin will support further analysis of profitability across products, categories, regions, customers, and discount levels.

For overall or grouped profit margin calculations, total Profit will be divided by total Sales rather than simply averaging individual transaction-level profit margins.

```python
df_clean["profit_margin"] = (
    df_clean["profit"] / df_clean["sales"]
)
```

```python
df_clean[
    ["sales", "profit", "profit_margin"]
].head(10)
```

### 4.9 Discount Impact Analysis

The previous analysis identified several loss-making Sub-Categories, particularly Tables, Bookcases, and Supplies. This section investigates whether discounting may be contributing to reduced profitability.

The analysis groups transactions by Discount level and examines:

- Total Sales.
- Total Profit.
- Number of Orders.
- Average Profit per transaction.

The objective is to determine how profitability changes as Discount levels increase and identify Discount levels that may be associated with significant business losses.

It is important to note that this analysis identifies patterns and associations between Discount and Profit but does not, by itself, prove that Discount is the sole cause of losses.

```python
discount_performance = (
    df_clean
    .groupby("discount")
    .agg(
        total_sales=("sales", "sum"),
        total_profit=("profit", "sum"),
        total_orders=("order_id", "nunique"),
        average_profit=("profit", "mean")
    )
    .reset_index()
)

discount_performance = (
    discount_performance
    .sort_values("discount")
    .reset_index(drop=True)
)

discount_performance.round(2)
```

#### Discount Impact Findings

The Discount-level analysis reveals a clear relationship between increasing Discount levels and declining profitability.

- Transactions with **no Discount (0%)** generated approximately **$320,988 in Total Profit**, representing the largest positive contribution to overall profitability.
- A **10% Discount** generated approximately **$9,029 in Profit** and recorded the highest Average Profit of approximately **$96.06** per transaction line.
- Transactions discounted by **15% and 20%** remained profitable, with the 20% Discount level generating approximately **$90,337 in Total Profit**.
- Profitability becomes negative from the **30% Discount level onward**. Every observed Discount level between 30% and 80% generated an overall loss.
- The **70% Discount level** generated the largest Total Loss at approximately **-$40,075**, followed by the **80% Discount level** at approximately **-$30,539**.
- The **50% Discount level** recorded the worst Average Profit at approximately **-$310.70** per transaction line.

#### Key Business Insight

The results suggest a strong negative association between high Discount levels and profitability. Discounts of up to 20% were associated with positive overall Profit, while all observed Discount levels of 30% and above generated losses.

However, these findings should not be interpreted as proof that Discount alone causes the losses. Profitability may also be influenced by factors such as product type, product cost, region, and order characteristics.

The results nevertheless indicate that high Discount levels represent an important area for further investigation, particularly for loss-making Sub-Categories such as Tables and Bookcases.

### 4.10 Visualization of Discount Impact on Profit

To better understand the relationship between Discount levels and profitability, Total Profit is visualized across the different Discount levels observed in the dataset.

- From the **30% Discount level onward**, every observed Discount level generated a negative Total Profit.
- Although the size of the losses varies across higher Discount levels, profitability does not recover above zero.
- The largest overall loss is associated with the **70% Discount level**.

#### Key Insight

The visualization provides evidence of a strong negative association between higher Discount levels and profitability in this dataset. The transition from positive to negative Total Profit occurs between the **20% and 30% Discount levels**.

### 4.18 Investigation of Loss-Making States

The State-level analysis identified several geographic markets that generated an overall loss despite recording substantial Sales.

This section investigates the loss-making States in greater detail by examining the relationship between State, Discount levels, Sales, and Profit.

The objective is to determine whether higher Discounts may be contributing to the losses recorded in these geographic markets and identify the States and Discount levels associated with the largest financial losses.

```python
loss_states = state_performance[
    state_performance["total_profit"] < 0
]["state"].tolist()

state_discount_performance = (
    df_clean[
        df_clean["state"].isin(loss_states)
    ]
    .groupby(["state", "discount"])
    .agg(
        total_sales=("sales", "sum"),
        total_profit=("profit", "sum"),
        total_orders=("order_id", "nunique")
    )
    .reset_index()
)

state_discount_performance = (
    state_discount_performance
    .sort_values("total_profit")
    .reset_index(drop=True)
)

state_discount_performance.round(2)
```

### Findings

- Texas recorded the largest loss at the 80% Discount level, generating a loss of approximately $20,852 from $11,450 in Sales.

- Ohio also recorded a significant loss at the 70% Discount level, with a loss of approximately $13,172.

- Other major losses occurred in Illinois at an 80% Discount, Colorado at 70%, and Pennsylvania at 70%.

- Texas remained profitable at the 20% Discount level, generating approximately $8,226 in Profit, but became loss-making at Discount levels of 30% and above.

- A similar pattern can be observed across several other loss-making States, where lower Discount levels generated positive Profit while higher Discount levels resulted in losses.

- The results indicate that excessive Discounting is a major factor contributing to poor profitability in these States.

- Therefore, the losses recorded in States such as Texas, Ohio, and Pennsylvania appear to be influenced not simply by weak Sales, but by the level of Discounts applied to transactions.

### Key Insight

Higher Sales do not necessarily result in higher Profit when aggressive Discounts are applied. The business should review its Discount strategy, particularly for Discount levels above 20%, to reduce avoidable losses in underperforming States.

### Discount Impact on Loss-Making States

The previous analysis identified several States that generated overall losses. Further analysis showed that higher Discount levels were strongly associated with these negative Profit outcomes.

To better understand this relationship, the following visualization compares Total Profit across different Discount levels for the major loss-making States.

This helps identify the Discount levels at which profitability begins to decline and highlights States that are particularly affected by aggressive Discounting.

```python
major_loss_states = [
    "Texas",
    "Ohio",
    "Pennsylvania",
    "Illinois",
    "Colorado"
]

discount_state_plot = state_discount_performance[
    state_discount_performance["state"].isin(major_loss_states)
]

plt.figure(figsize=(12, 7))

for state in major_loss_states:
    state_data = discount_state_plot[
        discount_state_plot["state"] == state
    ].sort_values("discount")
    
    plt.plot(
        state_data["discount"] * 100,
        state_data["total_profit"],
        marker="o",
        label=state
    )

plt.axhline(0, linestyle="--")

plt.title("Profit by Discount Level for Major Loss-Making States")
plt.xlabel("Discount (%)")
plt.ylabel("Total Profit ($)")
plt.legend(title="State")
plt.grid(True)
plt.tight_layout()

plt.show()
```
