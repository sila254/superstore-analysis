### 1.5 Key Performance Indicators (KPIs)

The following KPIs will be used to evaluate Superstore's business performance:

| KPI | Formula | Purpose |
|---|---|---|
| Total Sales | Sum of Sales | Measures total revenue generated |
| Total Profit | Sum of Profit | Measures overall profitability |
| Total Orders | Count of unique Order IDs | Measures the number of distinct customer orders |
| Total Quantity Sold | Sum of Quantity | Measures the total number of units sold |
| Average Order Value (AOV) | Total Sales / Total Orders | Measures the average revenue generated per order |
| Overall Profit Margin | Total Profit / Total Sales × 100 | Measures how much profit is generated from each unit of sales |
| Average Discount | Sum of Discount / Number of transaction lines | Measures the average discount applied across transaction lines |
| Sales Growth Rate | (Current Period Sales - Previous Period Sales) / Previous Period Sales × 100 | Measures the percentage change in sales between periods |
| Profit Growth Rate | (Current Period Profit - Previous Period Profit) / Previous Period Profit × 100 | Measures the percentage change in profit between periods |
| Average Shipping Time | Sum of Shipping Days / Number of Orders | Measures the average time between order and shipment dates |
| Loss-Making Transaction Rate | Number of loss-making transaction lines / Total transaction lines × 100 | Measures the proportion of transaction lines generating losses |

### 2.7 Numerical Values and Potential Outliers

The numerical variables were examined using summary statistics and extreme-value analysis to identify unusually high or low values and determine whether they represent potential data-quality issues.

#### Sales Observations

- The maximum transaction-line Sales value is **22,638.48**, while the minimum is **0.44**.

- The highest-sales transaction was for a **Cisco TelePresence System** in the **Technology** category, generating **22,638.48 in Sales**.
- Most of the highest-sales transactions involve high-value products from the **Technology** category, particularly copiers and printing equipment. Some high-value **Office Supplies** products also appear among the largest transactions.
- The smallest sales transactions are mainly low-value **Office Supplies** products.

#### Profit and Loss Observations

- The highest transaction-line Profit recorded is **8,399.98**, generated by a **Canon imageCLASS 2200 Advanced Copier** in the **Technology** category with **17,499.95 in Sales** and no discount.
- The largest transaction-line loss is **-6,599.98**, generated by a **Cubify CubeX 3D Printer** in the **Technology** category. The transaction had **4,499.99 in Sales** and received a **70% discount**.
- Several of the highest-profit transactions are associated with **Technology products**, particularly copiers.
- The largest losses are concentrated among high-value products from the **Technology** and **Office Supplies** categories, with one **Furniture** transaction also appearing among the ten largest losses.

#### Discount Observations

- A noticeable pattern appears among the largest loss-making transactions: many received substantial discounts ranging from **40% to 80%**.
- The largest loss received a **70% discount**, while several other major losses received discounts of **50%, 70%, or 80%**.
- In contrast, many of the highest-profit transactions received **no discount**, although at least one highly profitable transaction remained profitable with a **20% discount**.
- The highest-sales transaction received a **50% discount** and generated a loss of **1,811.08**, showing that high sales do not necessarily result in high profit.
- These observations suggest a possible relationship between high discount levels and reduced profitability. However, the extreme transactions alone are not sufficient to prove that discounts cause losses. The relationship will be tested across the full dataset during further analysis.

#### Outlier Assessment

- The dataset contains extreme Sales and Profit values compared with typical transaction values.
- However, the extreme values appear to be associated with identifiable high-value products rather than obvious data-entry errors.
- These potential outliers will therefore be retained rather than automatically removed.
- Further analysis will determine how much these extreme transactions influence overall sales and profitability results.
