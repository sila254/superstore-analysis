### 1.5 Key Performance Indicators (KPIs)

The following KPIs will be used to evaluate Superstore's business performance:

| KPI | Formula | Purpose |
|---|---|---|
| Total Sales | Sum of Sales | Measures total revenue generated |
| Total Profit | Sum of Profit | Measures overall profitability |
| Total Orders | Count of unique Order IDs | Measures the number of distinct customer orders |
| Total Quantity Sold | Sum of Quantity | Measures the total number of units sold |
| Average Order Value (AOV) | Total Sales / Total Orders | Measures the average revenue generated per order |
| Overall Profit Margin | Total Profit / Total Sales × 100 | Measures how much profit is generated from each unit of sales |
| Average Discount | Sum of Discount / Number of transaction lines | Measures the average discount applied across transaction lines |
| Sales Growth Rate | (Current Period Sales - Previous Period Sales) / Previous Period Sales × 100 | Measures the percentage change in sales between periods |
| Profit Growth Rate | (Current Period Profit - Previous Period Profit) / Previous Period Profit × 100 | Measures the percentage change in profit between periods |
| Average Shipping Time | Sum of Shipping Days / Number of Orders | Measures the average time between order and shipment dates |
| Loss-Making Transaction Rate | Number of loss-making transaction lines / Total transaction lines × 100 | Measures the proportion of transaction lines generating losses |

### 1.9 Initial Observations and Assumptions

#### Initial Observations

- The dataset contains 9,994 transaction-line records and 21 columns.
- The dataset is structured at the order-line level, meaning one order may appear across multiple rows when it contains multiple products.
- The data contains information about orders, customers, products, geographic locations, shipping, sales, discounts, quantities, and profit.
- The dataset covers multiple geographic regions and product categories.
- Sales and Profit are provided, but there is no separate cost-of-goods column.
- Order Date and Ship Date can be used to analyze sales trends and shipping times.
- The raw CSV requires explicit character encoding when loaded into Pandas.

#### Initial Assumptions

- Each Order ID represents a unique customer order.
- Each Customer ID represents a unique customer.
- Each Product ID represents a product identifier.
- Sales represents revenue generated by a transaction line.
- Profit represents the profit associated with a transaction line.
- Discount is represented as a decimal value, where 0.20 represents a 20% discount.
- Negative profit values represent loss-making transaction lines.
- The analysis will treat the provided data as historical business transaction data.
- These assumptions will be validated where possible during the data quality assessment.
